```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("IPL2025 Batters.csv")

# Display dataset
print(df.head())
print(df.info())
```

```python
top10 = df.sort_values("Runs", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.bar(top10["Player Name"], top10["Runs"])
plt.xlabel("Player Name")
plt.ylabel("Runs")
plt.title("Top 10 IPL 2025 Players by Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
team_count = df["Team"].value_counts()

plt.figure(figsize=(8,8))
plt.pie(
    team_count.values,
    labels=team_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("IPL 2025 Players Distribution by Team")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["Runs"], bins=10, edgecolor="black")
plt.xlabel("Runs")
plt.ylabel("Number of Players")
plt.title("Distribution of IPL 2025 Player Runs")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(12,8))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("IPL 2025 Batters Correlation Heat Map")
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df["Runs"], df["SR"], alpha=0.7)
plt.xlabel("Runs")
plt.ylabel("Strike Rate")
plt.title("Runs vs Strike Rate - IPL 2025")
plt.grid(True)
plt.show()
```
